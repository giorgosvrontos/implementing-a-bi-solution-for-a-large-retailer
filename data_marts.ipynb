```python
%sql
-- Data mart: συνολικά έσοδα/κέρδος ανά κανάλι πώλησης (POS vs Online)
CREATE OR REPLACE VIEW workspace.superstore.mart_revenue_by_channel AS
SELECT channel, SUM(sales_price) as total_revenue, SUM(profit) as total_profit, COUNT(*) as num_transactions
FROM workspace.superstore.gold_fact_sales
GROUP BY channel;
```

```python
%sql
-- Data mart: προϊόντα που χρειάζονται άμεση αναπλήρωση αποθέματος
CREATE OR REPLACE VIEW workspace.superstore.mart_stockout_risk AS
SELECT product_id, stock_quantity, restock_limit, needs_restock
FROM workspace.superstore.gold_fact_inventory
WHERE needs_restock = true;
```

```python
%sql
-- Data mart: πωλήσεις ανά μήνα, με βάση το dim_date
CREATE OR REPLACE VIEW workspace.superstore.mart_sales_by_month AS
SELECT 
    d.year,
    d.month,
    SUM(f.sales_price) as total_revenue,
    SUM(f.profit) as total_profit,
    COUNT(*) as num_transactions
FROM workspace.superstore.gold_fact_sales f
JOIN workspace.superstore.gold_dim_date d ON f.date_id = d.date_id
GROUP BY d.year, d.month
ORDER BY d.year, d.month;
```

```python
%sql
-- Data mart: πωλήσεις ανά κατηγορία προϊόντος (top-level)
CREATE OR REPLACE VIEW workspace.superstore.mart_sales_by_category AS
SELECT 
    p.category,
    SUM(f.sales_price) as total_revenue,
    SUM(f.profit) as total_profit,
    COUNT(*) as num_transactions
FROM workspace.superstore.gold_fact_sales f
JOIN workspace.superstore.gold_dim_product p ON f.product_id = p.product_id
GROUP BY p.category
ORDER BY total_revenue DESC;
```

```python
%sql
-- Data mart: πωλήσεις ανά υποκατηγορία προϊόντος (λεπτομερές)
CREATE OR REPLACE VIEW workspace.superstore.mart_sales_by_subcategory AS
SELECT 
    p.sub_category,
    p.category,
    SUM(f.sales_price) as total_revenue,
    SUM(f.profit) as total_profit,
    COUNT(*) as num_transactions
FROM workspace.superstore.gold_fact_sales f
JOIN workspace.superstore.gold_dim_product p ON f.product_id = p.product_id
GROUP BY p.sub_category, p.category
ORDER BY total_revenue DESC;
```
