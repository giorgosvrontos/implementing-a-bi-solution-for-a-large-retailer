```python
from pyspark.sql.functions import col

dim_product = spark.table("dbacademy.Superstore.sales_online_raw") \
    .select(
        col("Product ID").alias("product_id"),
        col("Product Name").alias("product_name"),
        col("Category").alias("category"),
        col("Sub-Category").alias("sub_category")
    ).distinct()

dim_product.write.mode("overwrite").saveAsTable("dbacademy.Superstore.dim_product")
```

```python
%sql
SELECT COUNT(*) FROM dbacademy.Superstore.dim_product;
```

```python
%sql
SELECT product_id, COUNT(*) as cnt
FROM dbacademy.Superstore.dim_product
GROUP BY product_id
HAVING COUNT(*) > 1
ORDER BY cnt DESC
LIMIT 10;
```
