```python
from pyspark.sql.functions import col, to_timestamp, date_format

pos_silver = spark.table("workspace.superstore.pos_sales_raw") \
    .withColumn("transaction_timestamp", to_timestamp(col("sale_timestamp"))) \
    .withColumn("date_id", date_format(col("transaction_timestamp"), "yyyyMMdd").cast("int")) \
    .select(
        col("order_id").alias("receipt_id"),
        col("date_id"),
        col("postal_code"),
        col("product_id"),
        col("customer_id"),
        col("channel"),
        col("transaction_timestamp"),
        col("quantity"),
        col("sales").alias("sales_price"),
        col("discount"),
        (col("sales") * 0.3).alias("profit")
    )

pos_silver.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("workspace.superstore.fact_sales_staging_pos")
```

```python
%sql
DESCRIBE workspace.superstore.fact_sales_staging_pos;
```
